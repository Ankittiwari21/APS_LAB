```python
import matplotlib.pyplot as plt
import numpy as np

# Data based on user input
categories = ['P(X=0)\nNo Click', 'P(X=1)\nClick']
theoretical = [0.70, 0.30]
empirical = [0.60, 0.40]

x = np.arange(len(categories))
width = 0.35

# Plotting
fig, ax = plt.subplots(figsize=(8, 5))
rects1 = ax.bar(x - width/2, theoretical, width, label='Theoretical', color='skyblue', edgecolor='black')
rects2 = ax.bar(x + width/2, empirical, width, label='Empirical', color='salmon', edgecolor='black')

# Formatting
ax.set_ylabel('Probability')
ax.set_title('Step 7: Theoretical vs Empirical PMF (Ad Click)')
ax.set_xticks(x)
ax.set_xticklabels(categories)
ax.set_ylim(0, 0.9)
ax.legend()

# Display values on top of bars
ax.bar_label(rects1, padding=3)
ax.bar_label(rects2, padding=3)

plt.tight_layout()
plt.savefig('ad_click_pmf.png')
print("Graph saved as ad_click_pmf.png")
```

```python
import numpy as np
import pandas as pd

# Define parameters
p = 0.3
sample_sizes = [10, 100, 1000, 10000]

# Theoretical values
theoretical_mean = p
theoretical_var = p * (1 - p)

# --- Step 13: Effect of sample size ---
results = []
for n in sample_sizes:
    # Generate empirical data (Bernoulli trials)
    data = np.random.binomial(1, p, n)

    # Calculate empirical metrics
    emp_p1 = np.mean(data)
    emp_p0 = 1 - emp_p1
    emp_mean = np.mean(data)
    emp_var = np.var(data) # Population variance for the sample

    results.append([n, emp_p0, emp_p1, emp_mean, emp_var])

df_step13 = pd.DataFrame(results, columns=[
    "Sample size", "Empirical P(X=0)", "Empirical P(X=1)",
    "Empirical mean", "Empirical variance"
])

print("--- Step 13 Table ---")
print(df_step13.to_string(index=False))
print("\n")

# --- Step 12: Final Result Table (using n=10000 as the final result) ---
final_n = 10000
final_data = np.random.binomial(1, p, final_n)
final_emp_mean = np.mean(final_data)
final_emp_var = np.var(final_data)

step12_data = [
    ["Expectation", theoretical_mean, final_emp_mean, abs(theoretical_mean - final_emp_mean)],
    ["Variance", theoretical_var, final_emp_var, abs(theoretical_var - final_emp_var)]
]

df_step12 = pd.DataFrame(step12_data, columns=[
    "Measure", "Theoretical value", "Empirical value", "Absolute difference"
])

print("--- Step 12 Table (n=10,000) ---")
print(df_step12.to_string(index=False))
```

```python
import numpy as np

# 1. Define the parameters
n = 50          # Number of transactions examined per hour
p = 0.04        # Probability of a transaction being flagged
hours = 20000   # Number of one-hour periods to simulate

# 2. Run the simulation
# np.random.binomial draws samples from a binomial distribution.
# It returns an array of length 20,000 containing the values of X for each hour.
X_simulated = np.random.binomial(n, p, hours)

# 3. View the results
print("First 15 simulated hours (number of flagged transactions):")
print(X_simulated[:15])

print(f"\nAverage flagged transactions per hour (Empirical Mean): {np.mean(X_simulated):.4f}")
print(f"Theoretical Mean (n * p): {n * p}")
print(f"Maximum flagged in a single hour: {np.max(X_simulated)}")
```
